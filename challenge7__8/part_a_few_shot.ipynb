**Group 8 | SVHN → MNIST | Digit Recognition**

Compares three strategies:
1. **Frozen backbone** (feature extraction)
2. **Fine-tuned** (layer3 + layer4 + fc unfrozen)

All trained on 50 labelled SVHN images per class (500 total), evaluated on SVHN test (source) and MNIST test (target).

```python
import sys, os
sys.path.insert(0, '..')   # find classifier.py
os.chdir('..')

from classifier import run_part_a

results, best_state = run_part_a(
    data_root='./data',
    figures_dir='./figures',
    ckpt_dir='./checkpoints',
    seeds=[42, 123, 7],
)
```

```python
import numpy as np

print(f'{'Model':<20} {'Src Acc':>12} {'Tgt Acc':>12} {'Δshift':>12}')
print('-' * 58)
for label, vals in results.items():
    s = np.array(vals['src_acc'])
    t = np.array(vals['tgt_acc'])
    d = s - t
    print(f'{label:<20} {s.mean():.4f}±{s.std():.4f}  {t.mean():.4f}±{t.std():.4f}  {d.mean():.4f}±{d.std():.4f}')
```
